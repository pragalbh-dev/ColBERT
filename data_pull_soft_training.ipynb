```python
import singlestoredb as s2
import pandas as pd
def fetch_company_details():
    """
    Fetches company details from the database
    """
    # Connect to the database
    conn = s2.connect(host='primary-singlestore.prod.synaptic.services',
                     port=3306,password='ManagerMemSQL@synaptic_production',
        user='manager_user',
                     database='labs_core_reloaded_production',connect_timeout=30,results_type='dict')
    cursor = conn.cursor()
    query = """
        select * from companies limit 20
    """
    cursor.execute(query)
    results = cursor.fetchall()
    return pd.DataFrame(results)
```
